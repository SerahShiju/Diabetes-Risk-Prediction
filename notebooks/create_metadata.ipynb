```python
# ============================================================
# Create Metadata File
# ============================================================

import json
import os

os.makedirs("../models", exist_ok=True)
```

```python
# ============================================================
# Load Saved Metadata Components
# ============================================================

import joblib
import pandas as pd

feature_names = joblib.load("../models/feature_names.pkl")

best_threshold = joblib.load("../models/threshold.pkl")

DATA_PATH = "../data/train.csv"

data = pd.read_csv(DATA_PATH)

TARGET = "diagnosed_diabetes"

X = data.drop(columns=[TARGET, "id"])

categorical_features = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()
```

```python
# ============================================================
# Build Metadata Dictionary
# ============================================================

metadata = {

    "project_name": "Diabetes Risk Prediction",

    "model_type": "Calibrated Stacking Ensemble",

    "target": "diagnosed_diabetes",

    "positive_class": 1,

    "negative_class": 0,

    "positive_label": "Diabetic",

    "negative_label": "Non-Diabetic",

    "threshold": float(best_threshold),

    "features": feature_names,

    "categorical": categorical_features,

    "numerical": numerical_features,

    "n_features": len(feature_names),

    "preprocessing": {

        "numerical": "Median Imputation + StandardScaler",

        "categorical": "Most Frequent Imputation + OneHotEncoder"

    },

    "base_models": [

        "XGBoost",

        "LightGBM",

        "MLP"

    ],

    "meta_model": "Logistic Regression",

    "calibration": "Isotonic Regression"

}
```

```python
# ============================================================
# Save Metadata
# ============================================================

with open("../models/metadata.json", "w") as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

print("metadata.json created successfully.")
```

```python
# ============================================================
# Verify Metadata
# ============================================================

with open("../models/metadata.json", "r") as file:

    metadata = json.load(file)

print(json.dumps(metadata, indent=4))
```
